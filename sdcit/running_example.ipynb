This package implements HSIC, KCIT, and SDCIT.
KCIT relies on `tensorflow` and `gpflow`, which we have tested with versions `tensorflow<2.0` and `gpflow<2.0` 


```python
import numpy as np
from sdcit.kcit import python_kcit, python_kcit_K
from sdcit.utils import rbf_kernel_median
from sdcit.sdcit_mod import SDCIT
import tensorflow as tf
tf.get_logger().setLevel('ERROR')
np.random.seed(0)

N = 200
print('jointly independent')
# Three independent random variables
X = np.random.randn(N, 2)
Y = np.random.randn(N, 2)
Z = np.random.randn(N, 2)
Kx, Ky, Kz = rbf_kernel_median(X, Y, Z)  # median heuristic
print(f'SDCIT p_value: {SDCIT(Kx, Ky, Kz)[1]:.4f}')
print(f'KCIT  p_value: {python_kcit(X, Y, Z)[2]:.4f}') 
print(f'KCIT  p_value: {python_kcit_K(Kx, Ky, Kz)[2]:.4f}')


print('\nconditionally dependent, X not _||_ Y | Z')
# (conditionally dependent)
# X --> Z <-- Y 
Z = X + Y + np.random.randn(N, 2)
Kx, Ky, Kz = rbf_kernel_median(X, Y, Z)  # median heuristic
print(f'SDCIT p_value: {SDCIT(Kx, Ky, Kz)[1]:.4f}')
print(f'KCIT  p_value: {python_kcit(X, Y, Z)[2]:.4f}') 
print(f'KCIT  p_value: {python_kcit_K(Kx, Ky, Kz)[2]:.4f}')


print('\nconditionally independent, X _||_ Y | Z')
# (conditionally independent)
# X <-- Z --> Y 
Z = np.random.randn(N, 2)
X = Z + np.random.randn(N, 2)
Y = Z + np.random.randn(N, 2)
Kx, Ky, Kz = rbf_kernel_median(X, Y, Z)  # median heuristic
print(f'SDCIT p_value: {SDCIT(Kx, Ky, Kz)[1]:.4f}')
print(f'KCIT  p_value: {python_kcit(X, Y, Z)[2]:.4f}') 
print(f'KCIT  p_value: {python_kcit_K(Kx, Ky, Kz)[2]:.4f}')
```
